```python

print(hotel.isnull().sum()[hotel.isnull().sum() > 0])


hotel['children'] = hotel['children'].fillna(0)


hotel['country'] = hotel['country'].fillna('Unknow')

hotel['agent'] = hotel['agent'].fillna(0)

#
hotel['company'] = hotel['company'].fillna(0)

print("\n--- Cleaning ke baad ---")
print(hotel.isnull().sum()[hotel.isnull().sum() > 0])
print("Ho gaya! Ab koi null nahi bacha")
```

```python

print("Duplicates:", hotel.duplicated().sum())
hotel = hotel.drop_duplicates()
print("Duplicates hatane ke baad shape:", hotel.shape)

print("\nGalat rows (0 guests):", ((hotel['adults']==0) & (hotel['children']==0) & (hotel['babies']==0)).sum())


hotel = hotel[~((hotel['adults']==0) & (hotel['children']==0) & (hotel['babies']==0))]
print("Saaf karne ke baad shape:", hotel.shape)

hotel['reservation_status_date'] = pd.to_datetime(hotel['reservation_status_date'])

```

```python

print("ADR 0 wali rows:", (hotel['adr']==0).sum())


hotel.loc[hotel['adr']==0, 'adr'] = hotel['adr'].median()


print("Lead time minus wali:", (hotel['lead_time']<0).sum())

# 4. 
print("\nFinal Nulls:")
print(hotel.isnull().sum().sum(), "nulls bache hain")

print("\nFinal Shape:", hotel.shape)
print("HOTEL CLEANING COMPLETE ✅")
```
